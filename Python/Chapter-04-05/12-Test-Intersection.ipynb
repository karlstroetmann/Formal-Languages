The function `regexp_2_string` converts a regular expression into a string that can be read back by the
function `parse` from the notebook `RegExp-Parser.ipynb`.

**Function `regexp_2_string(r)`**
- *Input:* `r` is a regular expression.
- *Output:* A string representing `r`.

```python
def regexp_2_string(r: Any) -> str:
    if r == 0: 
        return '∅'
    if r == '𝜀': 
        return '𝜀'
    if isinstance(r, str): # single characters
        return r
    if r[1] == '⋅': # concatenation
        r1, _, r2 = r
        return regexp_2_string(r1) + regexp_2_string(r2)        
    if r[1] == '+':
        r1, _, r2 = r
        return '(' + regexp_2_string(r1) + '+' + regexp_2_string(r2) + ')'
    if r[1] == '*': 
        r0 = r[0]
        if isinstance(r0, str):
            return regexp_2_string(r0) + '*'
        else:
            return '(' + regexp_2_string(r0) + ')*'
    raise Exception(f'{r} is not a suitable regular expression')
```

The function `test(Σ, s1, s2, expected)` computes the intersection of the regular expressions given by 
`s1` and `s2`.  In order to check the result, it uses the function `regExpEquiv` from the notebook 
`09-Equivalence.ipynb` to verify that the result is equivalent to the regular expression given by `expected`.

**Function `test(Σ, s1, s2, expected)`**
- *Input:* `Σ` is an alphabet and `s1`, `s2`, and `expected` are strings that represent regular expressions.
- *Output:* `None`.  The function prints the intersection of `s1` and `s2` and checks whether it is 
  equivalent to `expected`.

```python
def test(Σ: set[str], s1: str, s2: str, expected: str) -> None:
    r1 = parse(s1)
    r2 = parse(s2)
    r  = regexp_intersection(r1, r2, Σ)
    print(f'The intersection of {s1} and {s2} is {regexp_2_string(r)}.')
    if regExpEquiv(r, parse(expected), Σ):
        print(f'This is equivalent to {expected}, as expected.')
    else:
        print(f'ERROR: This is not equivalent to {expected}!')
```

```python
Σ = { "a", "b" }
```

The strings of even length that contain only the letter `a`:

```python
test(Σ, 'a*', '((a+b)(a+b))*', '(aa)*')
```

The strings that start with `a` and end with `b`:

```python
test(Σ, 'a(a+b)*', '(a+b)*b', 'a(a+b)*b')
```

The strings that contain both the substring `aa` and the substring `bb`:

```python
test(Σ, '(a+b)*aa(a+b)*', '(a+b)*bb(a+b)*', '(a+b)*(aa(a+b)*bb+bb(a+b)*aa)(a+b)*')
```

The intersection of disjoint languages is empty:

```python
test(Σ, 'a*', 'b(a+b)*', '∅')
```

The empty string is the only string that consists only of `a`s and only of `b`s:

```python
test(Σ, 'a*', 'b*', '𝜀')
```

The strings that contain an even number of `a`s and an even number of `b`s.  The 
product automaton has four states that keep track of the parity of the number of `a`s and the parity of the
number of `b`s.  The expected result is the well known regular expression 
$$ \bigl(aa + bb + (ab+ba)(aa+bb)^*(ab+ba)\bigr)^*. $$
The computed regular expression is equivalent to it, but it is much longer since the function `simplify`
does not know the equation $R^*R \doteq R^*$ that holds if $\varepsilon \in L(R)$.

```python
test(Σ, '(b*ab*a)*b*', '(a*ba*b)*a*', '(aa+bb+(ab+ba)(aa+bb)*(ab+ba))*')
```

The binary numbers that are divisible by $3$ and end with $0$, i.e. the binary numbers that are divisible

```python
test(Σ, '(0+1(01*0)*1)*', '(0+1)*0', '(0+1(01*0)*1)*0')
```
